### 2. Read the dataset

### 3. Configure Dataset

```python
print(df.columns) 
```

```python
# Split the first column into multiple columns
first_col_name = df.columns[0]  # original single column with CSV data
df = df[first_col_name].str.split(',', expand=True)

# Set proper column names
column_names = first_col_name.strip().split(',')
df.columns = column_names
df.head()
```

### 4. Data Assesments

#### 4.1 Check data types

```python
print(df.dtypes)
```

#### 4.1.1 Convert to numeric

```python
numeric_cols = ['tenure','MonthlyCharges','TotalCharges']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Handle NaNs if any
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

df.head()
```
