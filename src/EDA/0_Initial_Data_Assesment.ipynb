### 4. Data Cleaning & Export Processed Dataset
- Convert `TotalCharges` to numeric (handling 11 whitespace entries)
- Drop `customerID` (arbitrary identifier)
- Export cleaned data to `data/processed/Initial_Data_Assessed.csv`

```python
# 1. Convert TotalCharges to numeric, filling blank spaces with 0 (tenure is 0)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].astype(str).str.strip(), errors='coerce').fillna(0)

# 2. Drop customerID
if 'customerID' in df.columns:
    df = df.drop(columns=['customerID'])

# 3. Save to data/processed
save_dir = "../../data/processed"
os.makedirs(save_dir, exist_ok=True)
save_path = os.path.join(save_dir, "Initial_Data_Assessed.csv")
df.to_csv(save_path, index=False)
print(f"Cleaned dataset saved successfully: {save_path}")
print(f"Shape: {df.shape}")
```
